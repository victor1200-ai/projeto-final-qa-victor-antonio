```python
import pytest
import requests

def test_json_response():
    """Verifica se a resposta contém uma lista válida"""
    response = requests.get("https://reqres.in/api/users")
    data = response.json()
    assert isinstance(data["data"], list)
    assert len(data["data"]) > 0

# Executar `pytest` no Google Colab sem causar erro
pytest.main(["-q"])
```
